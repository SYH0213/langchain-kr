#### 네이버 개발자 센터에서 회원가입
https://developers.naver.com/main/

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
import requests
import json
import os

# 네이버 개발자 센터에서 발급받은 Client ID와 Secret
client_id = os.getenv("NAVER_CLIENT_ID")  # "YOUR_CLIENT_ID"
client_secret = os.getenv("NAVER_CLIENT_SECRET")  #

# 검색할 키워드
query = "상일동역 중화요리"

# API 요청 URL
url = f"https://openapi.naver.com/v1/search/local.json?query={query}&display=5"

# HTTP 헤더에 인증 정보 추가
headers = {"X-Naver-Client-Id": client_id, "X-Naver-Client-Secret": client_secret}

# API 요청 보내기
response = requests.get(url, headers=headers)
data = response.json()

# 결과 출력
print(json.dumps(data, indent=2, ensure_ascii=False))
```

```python
from langchain.tools import tool
import requests
from urllib.parse import quote

@tool
def search_shopping(query: str, display: int = 10, start: int = 1, sort: str = "sim", price_filter: str = None):
    """
    네이버 쇼핑 API에서 상품을 검색합니다.
    query: 검색어 (예: '아이폰 15')
    display: 검색 결과 개수 (기본 10, 최대 100)
    start: 시작 위치 (기본 1, 최대 1000)
    sort: 정렬 (sim=정확도, date=날짜순, asc=가격낮은순, dsc=가격높은순)
    price_filter: 'min:max' 형태의 가격 범위 (예: '500000:1500000')
    """
    
    
    # 1. 네이버 개발자 센터에서 발급받은 Client ID와 Secret
    client_id = os.getenv("NAVER_CLIENT_ID")  # "YOUR_CLIENT_ID"
    client_secret = os.getenv("NAVER_CLIENT_SECRET")  #

    # 2. 검색할 키워드
    search_query = "가성비 노트북"
    encoded_query = quote(search_query)  # 한글 검색어 URL 인코딩

    # 3. 요청 URL
    url = f"https://openapi.naver.com/v1/search/shop.json?query={encoded_query}&display={display}&start={start}&sort={sort}"

    # 4. 헤더 설정
    headers = {
        "X-Naver-Client-Id": client_id,
        "X-Naver-Client-Secret": client_secret
    }

    # 5. API 요청
    response = requests.get(url, headers=headers)

    # 6. 결과 확인
    if response.status_code == 200:
        data = response.json()
        for item in data["items"]:
            print(f"상품명: {item['title']}")
            print(f"가격: {item['lprice']}원")
            print(f"링크: {item['link']}")
            print("-" * 40)
    else:
        print("Error:", response.status_code, response.text)
```
