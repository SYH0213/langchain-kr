- Dog 클래스에는 name, breed 변수를 갖는다. 
- Dog 클래스에 어떤 변수와 속성을 갖는지 확인하는 방법으로 hasattr를 사용할 수 있다. 

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        return "Woof!"
```

```python
# 객체 생성
my_dog = Dog("Buddy", "Golden Retriever")

# Dog 객체에 속성 확인
print(f"my_dog에 'name' 속성이 있나요? {hasattr(my_dog, 'name')}")
print(f"my_dog에 'bark' 메서드가 있나요? {hasattr(my_dog, 'bark')}")
print(f"my_dog에 'age' 속성이 있나요? {hasattr(my_dog, 'age')}")
print(f"my_dog에 'age' 속성이 있나요? {hasattr(my_dog, 'breed')}")
```

```python
class Cat:
    def __init__(self, name):
        self.name = name
```

```python
my_cat = Cat("Whiskers")

# Cat 객체에 속성 확인
print(f"my_cat에 'name' 속성이 있나요? {hasattr(my_cat, 'name')}")
print(f"my_cat에 'breed' 속성이 있나요? {hasattr(my_cat, 'breed')}")
```

```python
my_string = "Hello, world!"

# 문자열 객체에 속성 확인
print(f"my_string에 'upper' 메서드가 있나요? {hasattr(my_string, 'upper')}")
print(f"my_string에 'append' 메서드가 있나요? {hasattr(my_string, 'append')}")
print(my_string.upper())
```
