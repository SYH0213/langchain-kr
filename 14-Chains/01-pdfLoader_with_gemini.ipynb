```python
from google import genai
from google.genai import types
import pathlib
import httpx

client = genai.Client()

# Retrieve and encode the PDF byte
file_path = pathlib.Path("data/gemini-2.5-tech_1-3.pdf")


# Upload the PDF using the File API
sample_file = client.files.upload(
    file=file_path,
)

prompt = """
            pdf파일에서 텍스트 내용을 추출해서 마크다운 파일을 생성할건데 다음 순서대로 진행해줘. 
            1. pdf에는 다양한 텍스트와 테이블, 차트가 포함되어 있어. 텍스트와 테이블만 추출해줘. 머리말과 꼬리말은 삭제해줘. 원문의 언어를 그대로 유지해줘
            2. 추출된 내용을 마크다운 파일로 생성해줘.
        """

response = client.models.generate_content(
    model="gemini-2.5-pro", contents=[sample_file, prompt]
)
print(response.text)
```
