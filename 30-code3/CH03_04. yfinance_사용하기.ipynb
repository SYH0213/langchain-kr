```python
import yfinance as yf

# Apple 티커 지정
aapl = yf.Ticker("AAPL")

# 지난 5일간의 주가 데이터
hist = aapl.history(period="5d")
print(hist)

# 재무제표 (분기별)
financials = aapl.financials
print(financials)

# 배당금 기록
dividends = aapl.dividends
print(dividends)
```
